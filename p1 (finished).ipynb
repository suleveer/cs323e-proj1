```python
import sympy as sp
import numpy as np

x = sp.symbols("x")

func_1 = (x - 4) * (x + 1)

func_2 = (x - 1) * (x + 3)

func_3 = (x - 4) * (x - 1) * (x + 3)

# non-trivial single-variable periodic function of our own
func_4 = (x**3 - (6 * (x**2)) - x + 25)
```

```python
def newton(f, guess, tol = 1.0e-9, max_iter = 200):

    f_prime = sp.diff(f, x, 1)
    
    for i in range(1, max_iter + 1):
        guess = (guess - (f.subs(x, guess) / f_prime.subs(x, guess))).evalf()
        
        if abs(f.subs(x, guess)) < tol:
            return guess, i

    return (guess, max_iter)
```

```python
# f1
vals = np.linspace(-10, 10, 500)
four = []
negative_one = []

for val in vals:
    root = newton(func_1, val)[0]
    if round(root) == 4:
        four.append(val)
    elif round(root) == -1:
        negative_one.append(val)

```

```python
# f1 plot

import matplotlib.pyplot as plt

# Plot the starting values according to the root they reached.
plt.figure(figsize=(12, 4))
plt.scatter(negative_one, np.zeros(len(negative_one)), s=12, label="root -1")
plt.scatter(four, np.zeros(len(four)), s=12, label="root 4")

# Plot a scaled version of func_1 on the same axes.
func_1_vals = np.array([float(func_1.subs(x, val)) for val in vals])
scaled_func_1 = func_1_vals / np.max(np.abs(func_1_vals))
plt.plot(vals, scaled_func_1, alpha=0.6, label="scaled F(x)")

plt.axvline(-1, alpha=0.3)
plt.axvline(4, color="tab:orange", alpha=0.3)
plt.xlim(-10, 10)
plt.xticks(np.arange(-10, 11, 2))
plt.gca().set_xticks(np.arange(-9, 10, 2), minor=True)
plt.tick_params(axis="x", which="minor", length=4)
plt.xlabel(r"starting value $x_0$")
plt.yticks([])
plt.title(r"Basins of attraction for $f_1(x)=(x-4)(x+1)$")
plt.legend()
plt.show()
```

```python
def secant(f, guess0, guess1, tol = 1.0e-9, max_iter = 200):
    
    for i in range(1, max_iter + 1):
        fx0 = f(guess0)
        fx1 = f(guess1)

        guess2 = guess1 - (fx1 * (guess1 - guess0) / (fx1 - fx0))
        
        if abs(f(guess2)) < tol:
            return guess2

        guess0 = guess1
        guess1 = guess2
    return (guess1, max_iter)

def func_3(x):
    return (x - 4) * (x - 1) * (x + 3)
```

```python
# f3 (secant)

vals = np.linspace(-10, 10, 500)
one = []
negative_three = []
four = []

for val in vals:
    root = secant(func_3, val, val + 1)
    if round(root) == 1:
        one.append(val)
    elif round(root) == -3:
        negative_three.append(val)
    elif round(root) == 4:
        four.append(val)
```

```python
# f3 plot (secant)

# Plot the starting values according to the root they reached.
plt.figure(figsize=(12, 4))
plt.scatter(negative_three, np.zeros(len(negative_three)), s=12, label="root -3")
plt.scatter(one, np.zeros(len(one)), s=12, label="root 1")
plt.scatter(four, np.zeros(len(four)), s=12, label="root 4")

# Plot a scaled version of func_3 on the same axes.
func_3_vals = np.array([float(func_3(val)) for val in vals])
scaled_func_3 = func_3_vals / np.max(np.abs(func_3_vals))
plt.plot(vals, scaled_func_3, alpha=0.6, label="scaled F(x)")

plt.axvline(-3, alpha=0.3)
plt.axvline(1, color="tab:orange", alpha=0.3)
plt.axvline(4, color="tab:green", alpha=0.3)
plt.ylim(-.5, .5)
plt.xlim(-10, 10)
plt.xticks(np.arange(-10, 11, 2))
plt.gca().set_xticks(np.arange(-9, 10, 2), minor=True)
plt.tick_params(axis="x", which="minor", length=4)
plt.xlabel(r"starting value $x_0$")
plt.yticks([])
plt.title(r"Basins of attraction for $f_3(x)=(x-4)(x-1)(x+3)$")
plt.legend()
plt.show()
```
